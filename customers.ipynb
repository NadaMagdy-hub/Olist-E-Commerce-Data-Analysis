-------  تنظيف وتوحيد نصوص أسماء المدن (إزالة المسافات الزائدة وتحويلها لحروف صغيرة)-------

```python
customers['customer_city'] = customers['customer_city'].astype(str).str.strip().str.lower()
```

------- تحويل الولايات لحروف كبيرة لتطابق تنسيق ملف الجيولكيشن بالظبط--------

```python
customers['customer_state'] = customers['customer_state'].astype(str).str.strip().str.upper()
```

------- الحفاظ على الـ Zip Code كـ Integer (رقم) ليطابق نوع بيانات ملف الجيولكيشن تماماً-------

```python
customers['customer_zip_code_prefix'] = customers['customer_zip_code_prefix'].astype(int)
```

---------فحص أبعاد الجدول والتحقق من أن معرف العميل فريد بالكامل--------

```python
print("Dataset Shape:", customers.shape)
print("Is customer_id fully unique?:", customers['customer_id'].nunique() == len(customers))
```

-------نحسب أولاً كم مرة اشترى كل عميل فريد------

```python
purchase_counts = customers['customer_unique_id'].value_counts()
```

--------نحدد العملاء الذين اشتروا أكثر من مرة-------

```python
repeat_customers = purchase_counts[purchase_counts > 1].count()
```

-------نحسب إجمالي عدد العملاء الفريدين-------

```python
total_unique_customers = customers['customer_unique_id'].nunique()
```

-----حساب النسبة المئوية-----

```python
repeat_rate = (repeat_customers / total_unique_customers) * 100
```
