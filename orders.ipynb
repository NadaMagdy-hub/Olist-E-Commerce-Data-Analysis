---------تحويل أعمدة التواريخ من نصوص إلى Datetime-----------

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors='coerce')
```

----------تنظيف النصوص وإزالة المسافات العشوائية---------

```python
orders['order_status'] = orders['order_status'].str.strip().str.lower()
orders['order_id'] = orders['order_id'].str.strip()
orders['customer_id'] = orders['customer_id'].str.strip()
```

-------- حذف الطلبات التي تم تعليمها كـ delivered ولكن ليس لها تاريخ توصيل فعلي (عددها 8 فقط)-------

```python
orders = orders.drop(orders[(orders['order_status'] == 'delivered') & (orders['order_delivered_customer_date'].isnull())].index)
```

------الكود ده بيعمل عمود جديد في جدول البيانات عشان يصنف الطلبات لـ "تم التوصيل" أو "لم يتم التوصيل" بناءً على حالة الطلب الحالية-----

```python
orders['delivery_status'] = orders['order_status'].apply(
    lambda x: 'Delivered' if x == 'delivered' else 'Not Delivered'
)
```

-------حسب لي بالظبط كم طلب طلع Delivered وكم طلب طلع Not Delivered-------

```python
orders['delivery_status'].value_counts()
```

------حساب الأيام المستغرقة للتوصيل الفعلي--------

```python
orders['delivery_time_days'] = (orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']).dt.days
```

---------حساب الأيام المتوقعة للتوصيل (حسب النظام)-------

```python
orders['estimated_delivery_time_days'] = (orders['order_estimated_delivery_date'] - orders['order_purchase_timestamp']).dt.days
```

------حساب الفرق بين التوصيل الفعلي والمتوقع (إذا كان السعر بالسالب يعني التوصيل متأخر)------

```python
orders['delivery_delta_days'] = (orders['order_estimated_delivery_date'] - orders['order_delivered_customer_date']).dt.days
```

-----إضافة مؤشر هل الطلب متأخر أم لا؟------
--الرقم 1 لو الطلب اتأخر فعلاً، والرقم 0 لو الطلب وصل في ميعاده أو بدري--

```python
orders['is_late'] = np.where(orders['delivery_delta_days'] < 0, 1, 0)
```
