```python
import pandas as pd

# 1. قراءة الملف الأصلي
df = pd.read_csv('cleaned_order_reviews.csv')

# 2. استبدال الأسطر الجديدة داخل التعليقات بمسافة حتى لا تكسر أسطر الـ CSV في MySQL
if 'review_comment_message' in df.columns:
    df['review_comment_message'] = df['review_comment_message'].astype(str).str.replace(r'[\r\n]+', ' ', regex=True)

if 'review_comment_title' in df.columns:
    df['review_comment_title'] = df['review_comment_title'].astype(str).str.replace(r'[\r\n]+', ' ', regex=True)

# 3. حفظ الملف المعدل
df.to_csv('final_order_reviews.csv', index=False, encoding='utf-8')

print("تم تنظيف الملف بنجاح وحفظه باسم final_order_reviews.csv!")
```
