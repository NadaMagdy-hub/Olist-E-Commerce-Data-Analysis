```python
import pandas as pd
```

```python
orders = pd.read_csv('cleaning_orders.csv')
order_items = pd.read_csv('cleaned_order_items.csv')
order_payments = pd.read_csv('cleaned_order_payments.csv')
reviews = pd.read_csv('cleaned_order_reviews.csv')
translation = pd.read_csv('cleaned_product_category_name_translation.csv')
```

----------- تنظيف وتصفية جداول البيانات (العناصر، المدفوعات، والتقييمات) لضمان اتساقها مع جدول الطلبات الرئيسي، مع تحديث جدول ترجمة فئات المنتجات لإضافة الفئات المفقودة.----------

```python
valid_order_ids = set(orders['order_id'])

order_items_fixed = order_items[order_items['order_id'].isin(valid_order_ids)]
order_payments_fixed = order_payments[order_payments['order_id'].isin(valid_order_ids)]
reviews_fixed = reviews[reviews['order_id'].isin(valid_order_ids)]

missing_translations = pd.DataFrame({
    'product_category_name': ['unknown', 'pc_gamer', 'portateis_cozinha_e_preparadores_de_alimentos'],
    'product_category_name_english': ['unknown', 'pc_gamer', 'kitchen_portables_and_food_preparators']
})
translation_fixed = pd.concat([translation, missing_translations], ignore_index=True)
```

```python
order_items_fixed.to_csv('cleaned_order_items.csv', index=False)
order_payments_fixed.to_csv('cleaned_order_payments.csv', index=False)
reviews_fixed.to_csv('cleaned_order_reviews.csv', index=False)
translation_fixed.to_csv('cleaned_product_category_name_translation.csv', index=False)
```
