-----تنظيف وتوحيد نصوص أسماء التصنيفات------

```python
Products['product_category_name'] = Products['product_category_name'].astype(str).str.strip().str.lower()
```

------معالجة القيم المفقودة في اسم التصنيف
بما إن pandas بيحول الـ NaN لـ 'nan' كـ string بعد الـ astype(str)، هنستبدلها بـ 'unknown'-----

```python
Products['product_category_name'] = Products['product_category_name'].replace('nan', 'unknown')
```

-----معالجة القيم المفقودة لأعمدة الخصائص الأخرى (الصور والأطوال)-------

```python
Products['product_name_lenght'] = Products['product_name_lenght'].fillna(0).astype('Int64')
Products['product_description_lenght'] = Products['product_description_lenght'].fillna(0).astype('Int64')
Products['product_photos_qty'] = Products['product_photos_qty'].fillna(0).astype('Int64')
```

--------معالجة الأوزان والأبعاد المفقودة (بالوسيط Median)-------

```python
Products['product_weight_g'] = Products['product_weight_g'].fillna(Products['product_weight_g'].median())
Products['product_length_cm'] = Products['product_length_cm'].fillna(Products['product_length_cm'].median())
Products['product_height_cm'] = Products['product_height_cm'].fillna(Products['product_height_cm'].median())
Products['product_width_cm'] = Products['product_width_cm'].fillna(Products['product_width_cm'].median())
```

```python
Products.to_csv('cleaned_products.csv', index=False)
```
