------أي خلية سابها العميل فاضية ومكتبش فيها تفاصيل، الكود بيملاها تلقائياً بكلمة: 'No Review' (بمعنى: لا يوجد نص تقييم).----

--------- إزالة المسافات الزائدة في البداية والنهاية وتحويل النصوص لحروف صغيرة (Lowercase)--------

```python
reviews['review_comment_title'] = reviews['review_comment_title'].astype(str).str.strip().str.lower()
reviews['review_comment_message'] = reviews['review_comment_message'].astype(str).str.strip().str.lower()
```

-------حساب الوقت المستغرق (بالأيام) بين تاريخ كتابة التقييم ورد المتجر عليه--------

```python
reviews['response_time_days'] = (reviews['review_answer_timestamp'] - reviews['review_creation_date']).dt.days
```

------- تصنيف التقييمات إلى (إيجابي / سلبي / محايد) بناءً على الـ review_score للتحليل السريع--------

```python
def classify_sentiment(score):
    if score >= 4:
        return 'positive'
    elif score == 3:
        return 'neutral'
    else:
        return 'negative'

reviews['sentiment'] = reviews['review_score'].apply(classify_sentiment)
```

-----استخراج سبب الشكوى تلقائياً من النص البرتغالي (للتعليقات السلبية)------

```python
def extract_complaint_reason(msg):
    msg = str(msg).lower()
    if 'não recebi' in msg or 'recebi' in msg or 'atraso' in msg or 'atrasado' in msg:
        return 'Delivery Delay / Non-Delivery'
    elif 'errado' in msg or 'outro produto' in msg:
        return 'Wrong Product Delivered'
    elif 'defeito' in msg or 'quebrado' in msg or 'estragado' in msg:
        return 'Damaged / Defective Product'
    elif 'falta' in msg or 'faltando' in msg:
        return 'Missing Items'
    elif msg == 'no message' or len(msg.strip()) == 0:
        return 'No Written Complaint'
    else:
        return 'Other Operational Issues'
```

----- تطبيق الدالة فقط على التقييمات السلبية، والإيجابية والمحايدة تاخد "Satisfied/Neutral Customer"-----

```python
reviews['complaint_reason'] = reviews.apply(
    lambda row: extract_complaint_reason(row['review_comment_message']) if row['sentiment'] == 'negative' else 'Satisfied/Neutral Customer',
    axis=1
)
```
